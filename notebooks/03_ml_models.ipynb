# Machine Learning Models
## Supermarket Analytics Assignment

This notebook demonstrates the machine learning models built for business insights including sales forecasting, promotional impact analysis, and store performance classification.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

# Add parent directory to path for imports
sys.path.append('..')
from src.supervised_learning import SupermarketAnalytics

# Set visualization style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
```

## Load Processed Data and Initialize Analytics

```python
# Initialize the analytics system
analytics = SupermarketAnalytics(data_path="../data/processed/supermarket_data_processed.csv")

# Load the processed data
analytics.load_data()

print(f"Loaded dataset shape: {analytics.data.shape}")
print(f"Dataset columns: {len(analytics.data.columns)}")
print("\nFirst few rows:")
print(analytics.data.head())
```

## Business Problem 1: Sales Forecasting

Can we predict sales amounts for promotional periods to improve inventory planning?

```python
# Run sales forecasting analysis
print("=" * 50)
print("SALES FORECASTING MODEL")
print("=" * 50)

# This will train and evaluate the sales forecasting model
forecasting_results = analytics.analyze_sales_forecasting()

print(f"\nModel Performance:")
print(f"R² Score: {forecasting_results['r2_score']:.4f}")
print(f"RMSE: {forecasting_results['rmse']:.2f}")
print(f"Mean Absolute Error: {forecasting_results['mae']:.2f}")

print(f"\nTop 5 Most Important Features:")
for feature, importance in forecasting_results['feature_importance'].head().items():
    print(f"  {feature}: {importance:.4f}")
```

## Business Problem 2: Promotional Impact Analysis

Which promotional strategies deliver the best return on investment?

```python
# Run promotional impact analysis
print("=" * 50)
print("PROMOTIONAL IMPACT ANALYSIS")
print("=" * 50)

promotion_results = analytics.analyze_promotion_impact()

print("\nPromotional Effectiveness Results:")
for strategy, metrics in promotion_results.items():
    if isinstance(metrics, dict):
        print(f"\n{strategy.replace('_', ' ').title()}:")
        if 'avg_sales' in metrics:
            print(f"  Average Sales: ${metrics['avg_sales']:.2f}")
        if 'uplift_percent' in metrics:
            print(f"  Uplift: {metrics['uplift_percent']:.1f}%")
        if 'sample_size' in metrics:
            print(f"  Sample Size: {metrics['sample_size']:,} transactions")
```

## Business Problem 3: Store Performance Classification

Can we identify characteristics that make stores high performers?

```python
# Run store performance classification
print("=" * 50)
print("STORE PERFORMANCE CLASSIFICATION")
print("=" * 50)

store_results = analytics.analyze_store_performance()

print(f"\nClassification Model Performance:")
print(f"Accuracy: {store_results['accuracy']:.4f}")
print(f"Precision: {store_results['precision']:.4f}")
print(f"Recall: {store_results['recall']:.4f}")
print(f"F1-Score: {store_results['f1_score']:.4f}")

print(f"\nStore Performance Distribution:")
if 'class_distribution' in store_results:
    for performance_class, count in store_results['class_distribution'].items():
        print(f"  {performance_class}: {count} stores")
```

## Model Evaluation and Business Insights

```python
# Generate comprehensive business insights
print("=" * 60)
print("COMPREHENSIVE BUSINESS INSIGHTS")
print("=" * 60)

insights = analytics.generate_business_insights()

for insight_name, insight_text in insights.items():
    print(f"\n{insight_name.upper().replace('_', ' ')}:")
    print(f"  {insight_text}")

print("\n" + "=" * 60)
```
